```python
import torch
```

```python
t1 = torch.tensor([1,2,3], requires_grad=True, dtype=torch.float32, device= 'cuda')
t2 = torch.tensor([3,4,5], requires_grad = True, dtype = torch.float32, device='cuda')
t3 = t1 * t2
t4 = 1/(1+torch.exp(t3))
t5 = t4.sum(); t5.retain_grad()
print(t5)
```

```python
t4.grad
```

```python
t5.backward()
```

```python
print(t5.grad)
```
