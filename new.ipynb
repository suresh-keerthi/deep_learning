```python
import numpy as np
```

```python
#takes sclers also
def sigmoid(arr):
    return 1/(1 + np.exp(-arr))

def dsigmoid(ele):
    return   sigmoid(ele) *(1 - sigmoid(ele))
```

```python
alpha = 0.3
```

```python
#one layer 
#2 neurons
# 2 inputs
#2 outputs
#for now let our l be sum of 2 y's
#for now ignore the bias

inputs = np.array([3,4]).reshape(2,1)
weights = np.random.random(4).reshape(2,2)  #each row one neuron
```

```python
#forward pass
inter_outputs =  weights @ inputs #shape = 2,1 
outputs = sigmoid(inter_outputs)
l = np.prod(outputs)
print("loss:",l)

#---------------------------
#udate weights ==>>
#trying to represent them as jacobains instead of gradinets or partial derivatives XD
dl_dy = np.array([ outputs[1, 0], outputs[0, 0] ]).reshape(1,2)
dy_dz = np.array([ [dsigmoid(inter_outputs[0, 0]), 0], [0, dsigmoid(inter_outputs[1, 0]) ] ]) #shae = 2,2
dz_dw = np.array([[inputs[0,0], inputs[1,0], 0, 0], [0, 0, inputs[0,0], inputs[1,0]] ]) # shape 2,4
#---------------------
# print(dl_dy)
# print("---------")
# print(dy_dz)
# print("---------")
# print(dz_dw)
#----------------------------------------
dl_dw = dl_dy  @ dy_dz @ dz_dw
print("dl_dw:" ,dl_dw)

#--------------------
#update weights
flattend_weights = weights.ravel() # this returns view so original weights will get updated, but doesn't matter here anyhow
flattend_weights = flattend_weights - (alpha * dl_dw.ravel())
weights = flattend_weights.reshape(2,2)
#--------------------------------
print( "weights:", weights)
```

```python
import torch
```

```python
Y = torch.tensor([[1], [0]], dtype=torch.float64)
W = torch.rand((2, 3), dtype=torch.float64, requires_grad=True)
X = torch.tensor([[1,0], [0,1]], dtype=torch.float64)
B = torch.tensor([0, 0, 0], dtype = torch.float64).resize_(1, 3); B.requires_grad = True
```

```python
Z = X @ W + B
A = torch.sigmoid(Z)
Yhat = torch.sum(A, dim=1, keepdim=True)
L = torch.square(Y - Yhat).sum(dim = 0, keepdim = True) * 1/X.shape[0]
print(Yhat)
print(L)
L.backward()
print(W.grad)
print(B.grad)
with torch.no_grad():
    W -= 0.01 * W
    B -= 0.01 * B
```
